## Step 1. DP 가우시안 메커니즘

임베딩을 공유하기 전에 (ε, δ)-DP 가우시안 노이즈를 더합니다. 표준(해석적) 가우시안
메커니즘의 노이즈 표준편차는

    sigma = sensitivity * sqrt(2 · ln(1.25/δ)) / ε

입니다. ε 이 작을수록 sigma 가 커져 보호가 강해집니다. 두 가지를 분명히 해 둡니다.

- 이 닫힌형 하한은 Dwork·Roth 의 고전적 결과로 **ε ∈ (0, 1) 에서만 엄밀한 DP 보장**입니다.
  아래에서 비교를 위해 ε ≥ 1 도 쓰지만, 그 구간에서는 "ε 에 반비례하는 노이즈 스케일"로만
  해석하고 정식 보장은 analytic Gaussian mechanism·Opacus 같은 회계 도구로 계산해야 합니다.
- `sigma = sensitivity/epsilon` 은 **Laplace 스케일이지 가우시안 표준편차가 아닙니다**(본문 §21.3.2).
  임베딩을 단위 노름으로 정규화(클리핑)했으므로 sensitivity=1.0 으로 둡니다.

## Step 3. "보호 강도"는 ε 값이 아니라 측정된 재식별 난이도로

ε 숫자 자체를 "강한 보호"라고 부르면 오해를 부릅니다 — 같은 ε 라도 데이터 분포·차원에 따라
실질 보호가 다릅니다(본문 §21.3.4). 그래서 라벨은 **위에서 측정한 재식별 성공률**로 붙입니다.

````python
def protection_label(reid_rate):
    # 무작위 추측 기준선 = 1/N_DOCS. 재식별 성공률로 보호 강도를 라벨링
    if reid_rate <= 3 / N_DOCS:
        return '강한 보호(재식별 거의 무작위 수준)'
    if reid_rate <= 0.35:
        return '중간 보호'
    return '약한 보호(재식별 쉬움)'

print(f'무작위 추측 재식별률 기준선: {1/N_DOCS:.3f}\n')
print(f"{'ε':>6}  {'재식별top1':>10}  {'Recall@5':>9}  보호 강도(측정 기반)")
print('-' * 60)
for eps in EPS_GRID:
    rec, reid = results[eps]
    print(f"{eps:6}  {reid:10.2f}  {rec:9.2f}  {protection_label(reid)}")
````

## 정리
- 원본 텍스트가 아니라 **DP 노이즈를 더한 임베딩만** 공유해 PII 노출 없이 공동 학습합니다.
- ε 은 유용성(Recall@k)과 재식별 난이도의 **정면 트레이드오프**를 조절합니다. 이 corpus·차원에서는
  작은 ε(강한 보호)에서 검색이 무너지고, 큰 ε 에서만 검색이 살아나되 재식별도 쉬워졌습니다 —
  고차원 임베딩에 per-벡터 DP 를 그대로 걸면 유용성 확보에 큰 ε 이 필요하다는 정직한 결과입니다.
- 그래서 "보호 강도"는 ε 숫자가 아니라 **측정된 재식별 성공률**로 라벨링해야 합니다.
- Federated 집계(여러 조직 기여 평균)는 per-조직 ε 를 고정한 채 노이즈를 √K 로 줄여 집계
  유용성을 회복시킵니다. hub 는 개별 원본을 보지 않고 각 기여는 여전히 DP 로 보호됩니다 —
  DP 와 secure aggregation 을 함께 쓰는 이유입니다.
- 정식 DP 보장(특히 ε ≥ 1)은 Opacus·dp-accounting 같은 검증된 회계 도구로 계산해야 합니다.
